```python
from langchain_core.messages import BaseMessage, HumanMessage
from langgraph.graph import END, MessageGraph
```

```python
def function1(message):
    return message

def function2(message):
    return message

def function3(message):
    return message

builder = MessageGraph()

builder.add_node("test1", function1)
builder.add_node("test2", function2)
builder.add_node("test3", function3)

builder.set_entry_point("test1")

builder.add_edge("test1", "test2")
builder.add_edge("test1", "test3")

builder.add_edge("test2", END)
builder.add_edge("test3", END)

graph = builder.compile()
```

```python
from IPython.display import Image, display
from langchain_core.runnables.graph import CurveStyle, MermaidDrawMethod, NodeStyles

from dotenv import load_dotenv

load_dotenv()

display(
    Image(
        graph.get_graph().draw_mermaid_png(
            draw_method=MermaidDrawMethod.API,
        )
    )
)
```
