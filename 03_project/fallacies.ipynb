```python
from typing import Annotated, List

from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END
from IPython.display import Image, display

from pydantic import BaseModel, Field

import operator

from langchain_openai import ChatOpenAI
```

```python
class Fallacy(BaseModel):
    name: str = Field(..., description="The name of the fallacy")
    description: str = Field(..., description="A description of the fallacy")
    example: str = Field(..., description="An example of the fallacy")

class Fallacies(BaseModel):
    fallacies: List[Fallacy] = Field(..., description="A list of fallacies")
```

```python
llm = ChatOpenAI(model="gpt-4o-mini")

planner = llm.with_structured_output(Fallacies)
```

```python
planner.invoke("Estou querendo criar 3 argumentos baseados em falacias")
```
